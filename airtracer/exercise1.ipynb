# Exercise 1: Where does the air come from?

In this exercise, we are going to study where the air over a region of your choice comes from. For this purpose, we are going to run a _backward_ simulation with FLEXPART.

In the [COMMAND](https://flexpart.img.univie.ac.at/docs/configuration.html#command) file:
- Set the direction of the simulation to backward.
- Select a simulation length of 6 days. The start and end times are up to you. Note that the end time in this case should be the time of the arrival, because it is a backward simulation.
- Set the output time step to 1 hour.

In the [RELEASES](https://flexpart.img.univie.ac.at/docs/configuration.html#releases) file:
- Set the time of the release to the end date of the simulation. Again this is needed because it is a backward simulation.
- Set the species to "air tracer".
- Choose a reasonable number of particles to release. More particles give a more representative picture but increase the simulation time. 100000 is a good number for our purpose.
- Set the release height to 2000 m (above ground).
- The release mass is irrelevant for backward simulations.
- Choose a release point or box of your choice.

In the [OUTGRID](https://flexpart.img.univie.ac.at/docs/configuration.html#outgrid) file:
- Select an output height that includes the whole troposphere, e.g. 15km. Only one height is needed for this exercise.
- Define the output grid centered on the release point. It grid should cover at least the area where you expect the air to come from, but should not be too big to save plotting time.

Submit the simulation. It will take a few minutes to finish. You can track the current simulation time in the runtime log.

Upload the gridded FLEXPART output (netcdf or binary) to the `flexpart_output` directory for plotting. If you have binary files, the output directory should also contain the namelists.

```python
import numpy as np
import sys
sys.path.append("../util/")
from IPython.display import HTML
from flexpart_io import *
from flexpart_plotting import *
```

## 1.1) Load the data

```python
# open flexpart
flxdir = "flexpart_output"
ds_nc, _ = open_flexpart_grid(flxdir, fileformat="netcdf")
```

## 1.2) Define colormap and levels

```python
# colormap and levels
cmap = plt.cm.BuPu

logmax = int(np.floor(np.log10(np.max(ds_nc["spec001_mr"]))))

levels = 10 ** np.arange(logmax - 7, logmax - 2 + 1e-10, 0.2)
norm = mcolors.LogNorm(vmin=10**(logmax-7), vmax=10**(logmax-2))
```

## 1.3) Plot the plume at a specific time step on a map
To see where the air comes from, we can plot the plume at a specific time step. Select the time step you want to plot (e.g. 120 corresponds to 5 days if you have hourly output).

```python
# select indices
timestep = 50
pointspec = [0]
nageclass = [0]
height = 0

# plot map
fig = plot_map(ds_nc.isel(time=[timestep], pointspec=pointspec, nageclass=nageclass, height=height), cmap, levels, norm)
```

**Notes on the result**: The unit "s" refers to the average time particles spend in each grid cell at the given time step in seconds. The area-integrated value corresponds to FLEXPART's output time step.

## 1.4) Plot animation of the plume
We can also plot an animation over multiple time steps. You can adapt the timesteps below. If you want to plot all timesteps you can set `timesteps = slice(None)`. This may take some time.

```python
# select indices
timesteps = slice(0, 72)
pointspec = [0]
nageclass = [0]
height = 0

# plot animation
ani = plot_map_anim(ds_nc.isel(time=timesteps, pointspec=pointspec, nageclass=nageclass), cmap, levels, norm)
HTML(ani.to_jshtml())
```
